#**Gestor de decisión para la concesión de un crédito**

```python
# Gestor de decisión para la concesión de un crédito

# Definición de variables
nombre = "Oscar"  # Nombre del solicitante
edad = 30  # Edad del solicitante
ingreso_mensual = 15000  # Ingreso mensual en pesos
deuda_mensual = 3000  # Pagos mensuales de otras deudas
historial_crediticio = True  # Indica si tiene buen historial crediticio
monto_solicitado = 100000  # Cantidad de crédito solicitada
antiguedad_laboral = 2  # Años de antigüedad en el empleo
pago_estimado = 5000  # Pago mensual estimado del nuevo crédito

# Cálculo de la capacidad de pago
ingreso_disponible = ingreso_mensual - deuda_mensual
# Para conocer el dinero disponible se resta la deuda mensual al ingreso mensual

porcentaje_deuda = (deuda_mensual / ingreso_mensual) * 100
# Se calcula el porcentaje del ingreso destinado a deudas.

# Evaluación de los requisitos del solicitante
edad_valida = edad >= 18 and edad <= 65
# Decisión: la edad debe ser mayor o igual a 18 y menor o igual a 65 años.

ingreso_valido = ingreso_mensual >= 10000
# Decisión: el ingreso mensual debe ser igual o mayor a $10000.

deuda_aceptable = porcentaje_deuda <= 40
# Decisión: el porcentaje de deuda actual no debe superar el 40%.

empleo_estable = antiguedad_laboral >= 1
# Decisión: el solicitante debe tener al menos un año de antigüedad laboral.

capacidad_pago = ingreso_disponible >= pago_estimado
# Decisión: el ingreso disponible debe cubrir el pago mensual estimado.

monto_valido = monto_solicitado > 0 and monto_solicitado <= 120000
# Decisión: el monto solicitado debe ser mayor que cero y no superar $120000.

# Decisión final sobre la concesión del crédito
if (edad_valida and ingreso_valido and deuda_aceptable
    and empleo_estable and capacidad_pago and monto_valido
    and historial_crediticio):
    # Decisión: se aprueba si todas las condiciones anteriores son verdaderas.
    print("Crédito aprobado para", nombre)
    print("Monto solicitado: $", monto_solicitado)
    print("Ingreso disponible: $", ingreso_disponible)
else:
    # Decisión: se rechaza si al menos una de las condiciones no se cumple.
    print("Crédito rechazado para", nombre)
    print("No se cumplen todos los requisitos.")

# Mostrar información financiera del solicitante
print("Ingreso mensual: $", ingreso_mensual)
print("Deuda mensual: $", deuda_mensual)
print("Porcentaje de deuda:", porcentaje_deuda, "%")
print("Antigüedad laboral:", antiguedad_laboral, "años")
print("Pago mensual estimado: $", pago_estimado)
```
